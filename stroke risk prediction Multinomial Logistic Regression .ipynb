```python
x = df.drop([
    "Patient_ID",
    "Stroke_Risk",
    "Stroke_Risk_Score",
    "AI_Health_Recommendation",
    "Doctor_Consultation_Needed"
], axis=1)

y = df["Stroke_Risk"]
```

```python
categorical_cols = [
    'Gender', 
    'Country', 
    'Smoking_Status', 
    'Alcohol_Consumption',
    'Physical_Activity_Level', 
    'Stress_Level',
    'Diet_Quality',
    'Family_History_Stroke',
    'Family_History_Heart_Disease',
    'Diabetes',
    'Hypertension',
    'Heart_Disease', 
    'Previous_TIA',
    'Atrial_Fibrillation',
    'Chronic_Kidney_Disease',
    'Medication_Adherence', 
    'Work_Type',
    'Residence_Type', 
    'Air_Pollution_Exposure'
]
```

```python
x = pd.get_dummies(x,columns=categorical_cols,drop_first=True,dtype=int)
```

```python
x.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.30, stratify = y, random_state = 42
)
```

```python
x_train.head()
```

```python
x_test.head()
```

```python
y_test.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
from sklearn.multiclass import OneVsRestClassifier
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = OneVsRestClassifier(
     LogisticRegression(
         max_iter = 1000,
         class_weight="balanced")
)
```

```python
model
```

```python
model.fit(x_train_scaled,y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
y_pred
```

```python
comparison = pd.DataFrame(
    {
        "Actual": y_test,
        "Predicted": y_pred
    }
)
comparison.head(15)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print("Accuracy :", accuracy_score(y_test,y_pred)*100)
```

```python
cm = confusion_matrix(y_test,y_pred)
```

```python
cm
```

```python
print("classification Report :\n", classification_report(y_test,y_pred))
```

**Conclusion:** The balanced Logistic Regression model achieved 81% accuracy with a 75% Macro F1-score. The model achieved 91% recall for High-risk and 83% recall for Low-risk, demonstrating improved identification of different stroke-risk categories despite class imbalance.

**Final metrics:**

- Accuracy: 81%
- Macro Precision: 71%
- Macro Recall: 84%
- Macro F1-Score: 75%
- Weighted F1-Score: 81%
